```python
from graphviz import Digraph

# --- 初始化 ---
dot = Digraph(
    'DA_Framework_Architecture',
    comment='Sequential DA Framework for HLM',
    graph_attr={
        'fontname': 'Helvetica', 
        'splines': 'ortho',
        'nodesep': '0.8',  # 略微增加节点间距
        'ranksep': '1.2',  # 略微增加层级间距
        'outputorder': 'edgesfirst',
        'compound': 'true'
    },
    node_attr={'shape': 'box', 'style': 'rounded,filled', 'fontname': 'Helvetica'},
    # 增大全局标签字体
    edge_attr={'fontname': 'Helvetica', 'fontsize': '10', 'arrowsize': '0.8'}
)
dot.attr(rankdir='TD', label='Sequential Data Assimilation Framework Architecture', fontsize='20', fontcolor='#333333')

# --- 1. 定义核心组件和数据结构 ---

# 使用一个不可见的表格来强制左右布局
with dot.subgraph(name='cluster_main_container') as container:
    container.attr(style='invis', rankdir='LR')

    # 左侧：主同化循环
    with container.subgraph(name='cluster_main_loop') as c:
        c.attr(label='Main Assimilation Loop (at Time Step t)', style='rounded,dashed', color='#a0a0a0', fontsize='14')
        c.node('analysis_ensemble_t', 'Analysis Ensemble\n(X_t|t)', shape='folder', fillcolor='#cde4ff', color='#5b9bd5')
        c.node('forecast_op', 'Forecast Operator (M)', fillcolor='#f8cbad', color='#ed7d31')
        c.node('forecast_ensemble_t1', 'Forecast Ensemble\n(X_t+1|t)', shape='folder', fillcolor='#e2efda', color='#70ad47')
        c.node('analysis_op', 'Analysis Operator (H)', fillcolor='#f8cbad', color='#ed7d31')
        c.node('simulated_obs', 'Simulated Observations\n(H(X_t+1|t))', shape='folder', fillcolor='#fce4d6', color='#ed7d31')
        c.node('real_obs', 'Real Observations\n(Y_t)', shape='folder', fillcolor='#fff2cc', color='#ffc000')
        c.node('kalman_update', 'Kalman Update Step', fillcolor='#f8cbad', color='#ed7d31')
        c.node('analysis_ensemble_t1', 'Updated Analysis Ensemble\n(X_t+1|t+1)', shape='folder', fillcolor='#cde4ff', color='#5b9bd5')

    # 右侧：外部数据源和服务组件
    with container.subgraph(name='cluster_externals') as c:
        c.attr(style='invis') 
        c.node('all_obs_source', 'Full Observation Timeseries Source', shape='box', style='rounded,filled', fillcolor='#f2f2f2', color='#808080')
        with c.subgraph(name='cluster_services') as services:
            services.attr(label='Persistent Services', style='rounded', color='#5a9bd5', fontsize='14')
            services.node('data_handler', 'Data Handler\n(State & Obs Manager)', shape='cylinder', fillcolor='#d9e1f2', color='#5a9bd5')
            services.node('hlm_runner', 'HLM Runner\n(Model Interface)', shape='cylinder', fillcolor='#d9e1f2', color='#5a9bd5')
        c.node('historical_states', 'Historical States DB\n{q_t-k|t-k...}', shape='cylinder', style='rounded,filled', fillcolor='#f2f2f2', color='#808080')

# --- 2. 创建数据流连接 ---

# 主循环内部数据流 (实线, 黑色)
dot.edge('analysis_ensemble_t', 'forecast_op', xlabel=' Input: X_t|t')
dot.edge('forecast_op', 'forecast_ensemble_t1', xlabel=' Output: X_t+1|t')
dot.edge('forecast_ensemble_t1', 'analysis_op', xlabel=' Input Ensemble')
dot.edge('analysis_op', 'simulated_obs', xlabel=' Output ')
dot.edge('forecast_ensemble_t1', 'kalman_update', xlabel=' Input Forecast Ensemble')
dot.edge('simulated_obs', 'kalman_update', xlabel=' Input Simulated Obs')
dot.edge('real_obs', 'kalman_update', xlabel=' Input Real Obs')
dot.edge('kalman_update', 'analysis_ensemble_t1', xlabel=' Output')

# 迭代循环 (虚线, 灰色) - 调整 port 使其更靠外
dot.edge('analysis_ensemble_t1', 'analysis_ensemble_t', 
         xlabel=' Updated Ensemble X_t+1|t+1\n(becomes next step\'s input) ', 
         style='dashed', color='#808080', constraint='false',
         tailport='e', headport='e', labeldistance='2.0', labelangle='-20')

# 服务调用 (虚线, 蓝色) - 增加 labeldistance 和 labelangle
dot.edge('forecast_op', 'hlm_runner', xlabel=' run_single_step()', style='dashed', color='#4472c4', constraint='false', labeldistance='1.5')
dot.edge('analysis_op', 'hlm_runner', xlabel=' run_window_simulation()', style='dashed', color='#4472c4', constraint='false', labeldistance='1.5')
dot.edge('analysis_op', 'historical_states', xlabel=' get_historical_analysis_state()', style='dashed', color='#4472c4', constraint='false', labeldistance='1.5')
dot.edge('data_handler', 'real_obs', xlabel=' get_observations(t)', style='dashed', color='#4472c4', labeldistance='1.5')
dot.edge('analysis_ensemble_t1', 'historical_states', xlabel=' store_analysis_state(t+1) ', style='dashed', color='#808080', constraint='false', labeldistance='1.5', headport='w', tailport='w')

# 数据存取 (点线, 黑色)
dot.edge('all_obs_source', 'data_handler', xlabel=' Loads at init ', style='dotted')
dot.edge('data_handler', 'historical_states', xlabel=' Stores/Retrieves ', style='dotted', dir='both')

# --- 渲染并保存图像 ---
output_filename = 'da_framework_architecture_final_tuned'
dot.render(output_filename, format='png', cleanup=True, view=False)

print(f"成功！DA 框架架构图已保存为 '{output_filename}.png'")
```
